```python
import numpy as np

# 1. Engineer repayment behavior features from train_prev before merging
prev_temp = train_prev.copy()
for col in ['approveddate', 'creationdate', 'closeddate', 'firstduedate', 'firstrepaiddate']:
    prev_temp[col] = pd.to_datetime(prev_temp[col], errors='coerce')

# Calculate days late (positive = late, negative = paid early) and loan duration
prev_temp['days_late'] = (prev_temp['firstrepaiddate'] - prev_temp['firstduedate']).dt.days
prev_temp['is_late'] = (prev_temp['days_late'] > 0).astype(int)
prev_temp['actual_duration'] = (prev_temp['closeddate'] - prev_temp['approveddate']).dt.days
prev_temp['prev_interest'] = prev_temp['totaldue'] - prev_temp['loanamount']

behavior_features = prev_temp.groupby('customerid').agg(
    avg_days_late=('days_late', 'mean'),
    max_days_late=('days_late', 'max'),
    late_payment_count=('is_late', 'sum'),
    late_payment_ratio=('is_late', 'mean'),
    avg_actual_duration=('actual_duration', 'mean'),
    avg_prev_interest=('prev_interest', 'mean')
).reset_index()

# Merge behavior features into master_df
master_df = master_df.merge(behavior_features, on='customerid', how='left')

# Fill NaNs for the 9 customers with no previous loans
behavior_cols = ['avg_days_late', 'max_days_late', 'late_payment_count', 
                 'late_payment_ratio', 'avg_actual_duration', 'avg_prev_interest']
master_df[behavior_cols] = master_df[behavior_cols].fillna(0)

# 2. Engineer features on the current loan in master_df
master_df['approveddate'] = pd.to_datetime(master_df['approveddate'])
master_df['creationdate'] = pd.to_datetime(master_df['creationdate'])
master_df['birthdate_dt'] = pd.to_datetime(master_df['birthdate'], errors='coerce')

# Customer Age at loan approval (fill missing ages with median age)
master_df['age'] = (master_df['approveddate'] - master_df['birthdate_dt']).dt.days / 365.25
master_df['age'] = master_df['age'].fillna(master_df['age'].median())

# Time features & Interest
master_df['approval_time_mins'] = (master_df['approveddate'] - master_df['creationdate']).dt.total_seconds() / 60.0
master_df['approved_hour'] = master_df['approveddate'].dt.hour
master_df['approved_dayofweek'] = master_df['approveddate'].dt.dayofweek
master_df['interest_amount'] = master_df['totaldue'] - master_df['loanamount']
master_df['interest_rate'] = master_df['interest_amount'] / master_df['loanamount']
master_df['was_referred'] = (master_df['referredby'] != 'Unknown').astype(int)

# Ratio of current loan to historical average
master_df['loan_to_prev_avg_ratio'] = np.where(
    master_df['previous_average_loan'] > 0,
    master_df['loanamount'] / master_df['previous_average_loan'],
    1.0
)

print("New master_df shape:", master_df.shape)
```

```python
from sklearn.model_selection import train_test_split

# Encode target: Good -> 1, Bad -> 0 (or vice versa depending on competition metric)
master_df['target'] = master_df['good_bad_flag'].map({'Good': 1, 'Bad': 0})

# Drop non-feature columns
drop_cols = [
    'customerid', 'systemloanid', 'approveddate', 'creationdate', 
    'birthdate', 'birthdate_dt', 'referredby', 'good_bad_flag', 
    'demo_match', 'previous_match', 'target'
]

X = master_df.drop(columns=drop_cols)
y = master_df['target']

# One-hot encode categorical variables
X = pd.get_dummies(X, drop_first=True)

# Stratified train-validation split (80% train, 20% validation)
X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f"Training features: {X_train.shape}, Validation features: {X_val.shape}")
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.metrics import accuracy_score, roc_auc_score, classification_report, ConfusionMatrixDisplay

models = {
    "Logistic Regression": make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000, class_weight='balanced')),
    "Random Forest": RandomForestClassifier(n_estimators=200, max_depth=10, class_weight='balanced', random_state=42),
    "Gradient Boosting": HistGradientBoostingClassifier(max_iter=200, learning_rate=0.05, max_depth=6, random_state=42)
}

for name, model in models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_val)
    probs = model.predict_proba(X_val)[:, 1]
    
    acc = accuracy_score(y_val, preds)
    auc = roc_auc_score(y_val, probs)
    
    print(f"=== {name} ===")
    print(f"Accuracy: {acc:.4f} | ROC-AUC: {auc:.4f}")
    print(classification_report(y_val, preds, target_names=['Bad (0)', 'Good (1)']))
```

```python
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, RocCurveDisplay

# 1. Define target: 1 = Bad Loan (Default - positive risk class), 0 = Good Loan
master_df['default_flag'] = master_df['good_bad_flag'].map({'Bad': 1, 'Good': 0})

# 2. Drop non-feature columns
drop_cols = [
    'customerid', 'systemloanid', 'approveddate', 'creationdate', 
    'birthdate', 'birthdate_dt', 'referredby', 'good_bad_flag', 
    'demo_match', 'previous_match', 'default_flag'
]

X = pd.get_dummies(master_df.drop(columns=drop_cols, errors='ignore'), drop_first=True)
y = master_df['default_flag']

# 3. Stratified Train-Validation Split (80% Train, 20% Validation)
X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 4. Initialize Models (with class balancing for Logistic Regression & Random Forest)
models = {
    "Logistic Regression": make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42)),
    "Random Forest": RandomForestClassifier(n_estimators=200, max_depth=8, class_weight='balanced', random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=150, learning_rate=0.05, max_depth=4, random_state=42)
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
results = []

plt.figure(figsize=(8, 6))
ax = plt.gca()

# 5. Train, Cross-Validate, and Evaluate
for name, model in models.items():
    cv_scores = cross_validate(model, X_train, y_train, cv=cv, scoring='roc_auc')
    
    model.fit(X_train, y_train)
    y_pred = model.predict(X_val)
    y_prob = model.predict_proba(X_val)[:, 1]
    
    results.append({
        "Model": name,
        "CV ROC-AUC (Mean ± Std)": f"{cv_scores['test_score'].mean():.3f} ± {cv_scores['test_score'].std():.3f}",
        "Val ROC-AUC": round(roc_auc_score(y_val, y_prob), 4),
        "Accuracy": round(accuracy_score(y_val, y_pred), 4),
        "Default Precision": round(precision_score(y_val, y_pred), 4),
        "Default Recall": round(recall_score(y_val, y_pred), 4),
        "Default F1-Score": round(f1_score(y_val, y_pred), 4)
    })
    
    RocCurveDisplay.from_predictions(y_val, y_prob, name=name, ax=ax)

plt.plot([0, 1], [0, 1], 'k--', label='Random Baseline (AUC = 0.50)')
plt.title("ROC Curve Comparison Across Models")
plt.legend(loc="lower right")
plt.show()

results_df = pd.DataFrame(results)
display(results_df)
```

```python
import joblib
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier

# 1. Select practical inputs a third-party user can enter in Streamlit
app_features = [
    'loanamount', 'totaldue', 'termdays', 'loannumber',
    'age', 'bank_account_type', 'employment_status_clients',
    'previous_loan_count', 'previous_average_loan',
    'late_payment_count', 'avg_days_late'
]

X_app = master_df[app_features].copy()
y_app = master_df['default_flag']

numeric_features = [
    'loanamount', 'totaldue', 'termdays', 'loannumber',
    'age', 'previous_loan_count', 'previous_average_loan',
    'late_payment_count', 'avg_days_late'
]
categorical_features = ['bank_account_type', 'employment_status_clients']

# 2. Build an end-to-end Pipeline (handles encoding & scaling automatically)
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ]
)

streamlit_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(n_estimators=200, max_depth=8, class_weight='balanced', random_state=42))
])

# 3. Fit on full dataset and save bundle
streamlit_pipeline.fit(X_app, y_app)

model_bundle = {
    'pipeline': streamlit_pipeline,
    'features': app_features,
    'account_types': sorted(X_app['bank_account_type'].unique().tolist()),
    'employment_statuses': sorted(X_app['employment_status_clients'].unique().tolist())
}

joblib.dump(model_bundle, 'credit_risk_model.pkl')
print("Saved 'credit_risk_model.pkl' successfully! Ready for Streamlit.")
```
